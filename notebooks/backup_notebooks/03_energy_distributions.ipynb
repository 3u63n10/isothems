# Notebook 03: Energy Distributions

```python
import numpy as np
import matplotlib.pyplot as plt
from utils_isotherms import (
    fE_sips,
    fE_toth,
    fE_jovanovic,
    fE_temkin,
    fE_freundlich,
    fE_DA,
    fE_hill
)
```

## Overlay of all distributions in reduced energy u = (E - Ec)/RT

```python
# Normalized parameters
T = 298.0
RT = 8.314e-3 * T  # kJ/mol
Ec = 0
Qmax = 1

# Energy array from -10*RT to 10*RT
E = np.linspace(-10*RT, 10*RT, 500)

# Compute distributions (signatures from utils_isotherms.py)
# fE_sips(E, Ec, n, Qmax, T)
# fE_toth(E, Ec, t, Qmax, T)
# fE_jovanovic(E, Ec, Qmax, T)
# fE_temkin(E, Emin, Emax, Qmax)
# fE_freundlich(E, n, K0, T)
# fE_hill(E, Ec, nH, Qmax, T)

f_sips = fE_sips(E, Ec, 1.0, Qmax, T)
f_toth = fE_toth(E, Ec, 0.5, Qmax, T)
f_jovanovic = fE_jovanovic(E, Ec, Qmax, T)
f_temkin = fE_temkin(E, -3*RT, 3*RT, Qmax)
f_freundlich = fE_freundlich(E, 0.5, 1.0, T)
f_hill = fE_hill(E, Ec, 2.0, Qmax, T)

# Clip Freundlich to avoid divergence at high E
f_freundlich[E >= 5*RT] = np.nan

# Plot
plt.figure(figsize=(8, 5))
u = (E - Ec) / RT
plt.plot(u, f_sips * RT / Qmax, label='Sips (n=1.0)', linewidth=2)
plt.plot(u, f_toth * RT / Qmax, label='Tóth (t=0.5)', linewidth=2)
plt.plot(u, f_jovanovic * RT / Qmax, label='Jovanovic', linewidth=2)
plt.plot(u, f_temkin * RT / Qmax, label='Temkin', linewidth=2)
plt.plot(u, f_freundlich * RT / Qmax, label='Freundlich (n=0.5)', linewidth=2)
plt.plot(u, f_hill * RT / Qmax, label='Hill (nH=2.0)', linewidth=2)
plt.xlabel('$(E - E_c)/RT$')
plt.ylabel('$f(E) \\cdot RT / Q_{\\max}$')
plt.legend()
plt.grid(True, alpha=0.3)
plt.title('Overlay of Normalised Energy Distributions')
plt.tight_layout()
plt.show()
```

## Parameter variation: Sips distribution for different m

```python
# Sips parameter variation: n (heterogeneity exponent)
n_values = [0.5, 0.7, 1.0, 1.5, 2.0]
colors_v = plt.cm.viridis(np.linspace(0, 1, len(n_values)))

plt.figure(figsize=(8, 5))
u = (E - Ec) / RT
for n_val, color in zip(n_values, colors_v):
    f = fE_sips(E, Ec, n_val, Qmax, T)
    plt.plot(u, f * RT / Qmax, label=f'n = {n_val}', linewidth=2, color=color)

plt.xlabel('$(E - E_c)/RT$')
plt.ylabel('$f(E) \\cdot RT / Q_{\\max}$')
plt.title('Sips Energy Distribution — Effect of n')
plt.legend()
plt.grid(True, alpha=0.3)
plt.tight_layout()
plt.show()
```

## sigma_E vs heterogeneity parameter

```python
# Sips: m from 0.3 to 3.0
m_range = np.linspace(0.3, 3.0, 100)
sigma_sips = np.pi * m_range / np.sqrt(3)

# Hill: nH from 0.3 to 4.0
nH_range = np.linspace(0.3, 4.0, 100)
sigma_hill = np.pi / (np.sqrt(3) * nH_range)

# Jovanovic: constant
sigma_jovanovic = np.pi / np.sqrt(6)

plt.figure(figsize=(8, 5))
plt.plot(m_range, sigma_sips, label='Sips: σ = πm/√3', linewidth=2)
plt.plot(nH_range, sigma_hill, label='Hill: σ = π/(√3·nH)', linewidth=2)
plt.axhline(sigma_jovanovic, color='green', linestyle='--', 
            label=f'Jovanovic: σ = π/√6 = {sigma_jovanovic:.3f}', linewidth=2)
plt.xlabel('Heterogeneity parameter')
plt.ylabel('sigma_E/RT')
plt.legend()
plt.grid(True, alpha=0.3)
plt.tight_layout()
plt.show()
```

## D-R/D-A Weibull distributions

```python
# D-R / D-A: Weibull distributions
# fE_DA(eps, Qmax, Ea, nDA)
Ea = 10  # kJ/mol
Qmax_DA = 1
eps = np.linspace(0.01, 30, 500)
nDA_values = [1.5, 2.0, 3.0, 4.0]

plt.figure(figsize=(8, 5))
for nDA in nDA_values:
    f = fE_DA(eps, Qmax_DA, Ea, nDA)
    plt.plot(eps, f, label=f'$n_{{DA}}$ = {nDA}', linewidth=2)

plt.xlabel('Polanyi potential $\\varepsilon$ (kJ/mol)')
plt.ylabel('$f(\\varepsilon)$')
plt.title('Dubinin–Astakhov Energy Distributions')
plt.legend()
plt.grid(True, alpha=0.3)
plt.tight_layout()
plt.show()
```
