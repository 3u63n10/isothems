# Notebook 01: Isotherm Fitting

This notebook demonstrates non-linear fitting of multiple adsorption isotherm models to synthetic data. We generate synthetic data from a known Langmuir isotherm, add realistic noise, and then fit various candidate models to evaluate their performance.

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.optimize import curve_fit
import pandas as pd

from utils_isotherms import (
    langmuir, freundlich, sips, toth, jovanovic, temkin, hill,
    dubinin_radushkevich, redlich_peterson, koble_corrigan,
    radke_prausnitz, khan, fritz_schlunder,
    generate_synthetic_data, MODELS,
    sse, rmse, mae, r_squared, aic, bic
)
```

## Generate synthetic Langmuir data

```python
# Generate synthetic data from Langmuir model
Qmax_true = 100
KL_true = 0.05
noise_level = 0.03
n_points = 20

Ce, q_exp, q_true = generate_synthetic_data(
    model_func=langmuir,
    params=[Qmax_true, KL_true],
    Ce_range=(0.1, 100),
    n_points=n_points,
    noise_level=noise_level,
    random_seed=42
)

# Plot the synthetic data
plt.figure(figsize=(8, 5))
plt.scatter(Ce, q_exp, color='blue', label='Synthetic data (with noise)', s=50, alpha=0.7)
plt.plot(Ce, q_true, color='red', linewidth=2, label='True Langmuir', linestyle='--')
plt.xlabel('Concentration (Ce)', fontsize=12)
plt.ylabel('Adsorption capacity (q)', fontsize=12)
plt.title('Synthetic Langmuir Data', fontsize=14)
plt.legend()
plt.grid(True, alpha=0.3)
plt.tight_layout()
plt.show()

print(f"True parameters: Qmax = {Qmax_true}, KL = {KL_true}")
print(f"Number of data points: {n_points}")
print(f"Noise level: {noise_level*100}%")
```

## Fit all candidate models

```python
# Define models to fit with their initial guesses and bounds
models_to_fit = [
    {
        'name': 'Langmuir',
        'func': langmuir,
        'p0': [80, 0.1],
        'bounds': ([0, 0], [500, 10])
    },
    {
        'name': 'Freundlich',
        'func': freundlich,
        'p0': [10, 0.5],
        'bounds': ([0, 0], [500, 1])
    },
    {
        'name': 'Sips',
        'func': sips,
        'p0': [80, 0.05, 0.8],
        'bounds': ([0, 0, 0.1], [500, 10, 2])
    },
    {
        'name': 'Toth',
        'func': toth,
        'p0': [80, 0.05, 0.8],
        'bounds': ([0, 0, 0.1], [500, 10, 2])
    },
    {
        'name': 'Jovanovic',
        'func': jovanovic,
        'p0': [80, 0.05],
        'bounds': ([0, 0], [500, 10])
    },
    {
        'name': 'Temkin',
        'func': temkin,
        'p0': [0.5, 0.1],
        'bounds': ([0, 0.001], [100, 100])
    },
    {
        'name': 'Hill',
        'func': hill,
        'p0': [80, 20, 1],
        'bounds': ([0, 0.01, 0.1], [500, 1e6, 5])
    }
]

# Fit each model and collect results
results = []

for model_info in models_to_fit:
    try:
        # Fit the model
        popt, pcov = curve_fit(
            model_info['func'],
            Ce,
            q_exp,
            p0=model_info['p0'],
            bounds=model_info['bounds'],
            maxfev=10000
        )
        
        # Calculate predictions
        q_pred = model_info['func'](Ce, *popt)
        
        # Calculate error metrics
        sse_val = sse(q_exp, q_pred)
        r2_val = r_squared(q_exp, q_pred)
        
        # Store results
        results.append({
            'model': model_info['name'],
            'params': popt,
            'SSE': sse_val,
            'R2': r2_val,
            'q_pred': q_pred,
            'func': model_info['func']
        })
        
        print(f"✓ {model_info['name']}: SSE = {sse_val:.4f}, R² = {r2_val:.4f}")
        
    except Exception as e:
        print(f"✗ {model_info['name']}: Failed to fit - {str(e)}")
        results.append({
            'model': model_info['name'],
            'params': None,
            'SSE': np.inf,
            'R2': -np.inf,
            'q_pred': None,
            'func': None
        })

print(f"\nSuccessfully fit {sum([1 for r in results if r['params'] is not None])}/{len(models_to_fit)} models")
```

## Results summary

```python
# Create DataFrame for results summary
results_df = pd.DataFrame([
    {
        'Model': r['model'],
        'SSE': r['SSE'],
        'R²': r['R2'],
        'Parameters': str(np.round(r['params'], 4)) if r['params'] is not None else 'Failed'
    }
    for r in results
])

# Sort by SSE (best fit first)
results_df = results_df.sort_values('SSE').reset_index(drop=True)

# Display results
print("Model Fitting Results (sorted by SSE):")
print("="*80)
display(results_df)

# Highlight best model
best_model = results_df.iloc[0]
print(f"\nBest model: {best_model['Model']} (SSE = {best_model['SSE']:.4f}, R² = {best_model['R²']:.4f})")
```

## Visualize best fits

```python
# Plot experimental data and top 4 model fits
plt.figure(figsize=(8, 5))

# Plot experimental data
plt.scatter(Ce, q_exp, color='black', label='Experimental data', s=60, alpha=0.7, zorder=5)

# Get top 4 models (sorted by SSE)
sorted_results = sorted(results, key=lambda x: x['SSE'])
top_4_results = [r for r in sorted_results if r['q_pred'] is not None][:4]

# Define colors for the top models
colors = ['red', 'blue', 'green', 'orange']

# Plot each of the top 4 models
for i, result in enumerate(top_4_results):
    # Generate smooth curve for plotting
    Ce_smooth = np.linspace(Ce.min(), Ce.max(), 200)
    q_smooth = result['func'](Ce_smooth, *result['params'])
    
    plt.plot(
        Ce_smooth,
        q_smooth,
        color=colors[i],
        linewidth=2,
        label=f"{result['model']} (SSE={result['SSE']:.3f})",
        alpha=0.8
    )

plt.xlabel('Concentration (Ce)', fontsize=12)
plt.ylabel('Adsorption capacity (q)', fontsize=12)
plt.title('Top 4 Isotherm Model Fits', fontsize=14)
plt.legend(loc='best', fontsize=10)
plt.grid(True, alpha=0.3)
plt.tight_layout()
plt.show()
```
