# Notebook 03: Energy Distributions

```python
import numpy as np
import matplotlib.pyplot as plt
from utils_isotherms import (
    fE_sips,
    fE_toth,
    fE_jovanovic,
    fE_temkin,
    fE_freundlich,
    fE_DA,
    fE_hill
)
```

## Overlay of all distributions in reduced energy u = (E - Ec)/RT

```python
# Normalized parameters
RT = 2.478
Ec = 0
Qmax = 1

# Energy array from -10*RT to 10*RT
E = np.linspace(-10*RT, 10*RT, 500)

# Compute distributions
f_sips = fE_sips(E, Ec, Qmax, RT, m=1.0)
f_toth = fE_toth(E, Ec, Qmax, RT, t=0.5)
f_jovanovic = fE_jovanovic(E, Ec, Qmax, RT)
f_temkin = fE_temkin(E, Qmax, Emin=-3*RT, Emax=3*RT)
f_freundlich = fE_freundlich(E, Qmax, RT, n=0.5, K0=1.0)
f_hill = fE_hill(E, Ec, Qmax, RT, nH=2.0)

# Clip Freundlich to E < 5*RT
f_freundlich[E >= 5*RT] = np.nan

# Plot
plt.figure(figsize=(8, 5))
u = (E - Ec) / RT
plt.plot(u, f_sips, label='Sips (n=1.0)', linewidth=2)
plt.plot(u, f_toth, label='Toth (t=0.5)', linewidth=2)
plt.plot(u, f_jovanovic, label='Jovanovic', linewidth=2)
plt.plot(u, f_temkin, label='Temkin', linewidth=2)
plt.plot(u, f_freundlich, label='Freundlich (n=0.5)', linewidth=2)
plt.plot(u, f_hill, label='Hill (nH=2.0)', linewidth=2)
plt.xlabel('(E - Ec)/RT')
plt.ylabel('f(E)·RT/Qmax')
plt.legend()
plt.grid(True, alpha=0.3)
plt.tight_layout()
plt.show()
```

## Parameter variation: Sips distribution for different m

```python
# Parameter values
m_values = [0.5, 0.7, 1.0, 1.5, 2.0]
colors = plt.cm.viridis(np.linspace(0, 1, len(m_values)))

plt.figure(figsize=(8, 5))
for m, color in zip(m_values, colors):
    f = fE_sips(E, Ec, Qmax, RT, m=m)
    plt.plot(u, f, label=f'm = {m}', linewidth=2, color=color)

plt.xlabel('(E - Ec)/RT')
plt.ylabel('f(E)·RT/Qmax')
plt.legend()
plt.grid(True, alpha=0.3)
plt.tight_layout()
plt.show()
```

## sigma_E vs heterogeneity parameter

```python
# Sips: m from 0.3 to 3.0
m_range = np.linspace(0.3, 3.0, 100)
sigma_sips = np.pi * m_range / np.sqrt(3)

# Hill: nH from 0.3 to 4.0
nH_range = np.linspace(0.3, 4.0, 100)
sigma_hill = np.pi / (np.sqrt(3) * nH_range)

# Jovanovic: constant
sigma_jovanovic = np.pi / np.sqrt(6)

plt.figure(figsize=(8, 5))
plt.plot(m_range, sigma_sips, label='Sips: σ = πm/√3', linewidth=2)
plt.plot(nH_range, sigma_hill, label='Hill: σ = π/(√3·nH)', linewidth=2)
plt.axhline(sigma_jovanovic, color='green', linestyle='--', 
            label=f'Jovanovic: σ = π/√6 = {sigma_jovanovic:.3f}', linewidth=2)
plt.xlabel('Heterogeneity parameter')
plt.ylabel('sigma_E/RT')
plt.legend()
plt.grid(True, alpha=0.3)
plt.tight_layout()
plt.show()
```

## D-R/D-A Weibull distributions

```python
# Parameters
Ea = 10  # kJ/mol
Qmax_DA = 1
eps = np.linspace(0, 30, 500)
nDA_values = [1.5, 2.0, 3.0, 4.0]

plt.figure(figsize=(8, 5))
for nDA in nDA_values:
    f = fE_DA(eps, Ea, Qmax_DA, nDA)
    plt.plot(eps, f, label=f'nDA = {nDA}', linewidth=2)

plt.xlabel('Polanyi potential eps (kJ/mol)')
plt.ylabel('f(eps)')
plt.legend()
plt.grid(True, alpha=0.3)
plt.tight_layout()
plt.show()
```
