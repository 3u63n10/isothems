# Notebook 02: Error Metrics and Model Selection

```python
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
from scipy.optimize import curve_fit
from utils_isotherms import (
    langmuir,
    sips,
    freundlich,
    generate_synthetic_data,
    all_error_metrics,
    compute_AICc,
    compute_BIC,
    akaike_weights,
    f_test
)
```

## Generate and fit synthetic data

```python
# Generate synthetic Sips data
C_exp, q_exp = generate_synthetic_data(
    model='sips',
    params={'Qmax': 120, 'KS': 0.08, 'm': 0.7},
    C_range=(0.1, 100),
    n_points=25,
    noise_level=0.05,
    seed=42
)

# Fit Langmuir model
popt_langmuir, _ = curve_fit(
    langmuir,
    C_exp,
    q_exp,
    p0=[100, 0.1],
    bounds=([0, 0], [500, 10])
)
q_pred_langmuir = langmuir(C_exp, *popt_langmuir)

# Fit Sips model
popt_sips, _ = curve_fit(
    sips,
    C_exp,
    q_exp,
    p0=[100, 0.08, 0.8],
    bounds=([0, 0, 0.1], [500, 10, 2])
)
q_pred_sips = sips(C_exp, *popt_sips)

# Fit Freundlich model
popt_freundlich, _ = curve_fit(
    freundlich,
    C_exp,
    q_exp,
    p0=[15, 0.4],
    bounds=([0, 0.01], [500, 1])
)
q_pred_freundlich = freundlich(C_exp, *popt_freundlich)

print(f"Langmuir parameters: Qmax={popt_langmuir[0]:.2f}, KL={popt_langmuir[1]:.4f}")
print(f"Sips parameters: Qmax={popt_sips[0]:.2f}, KS={popt_sips[1]:.4f}, m={popt_sips[2]:.4f}")
print(f"Freundlich parameters: KF={popt_freundlich[0]:.2f}, n={popt_freundlich[1]:.4f}")
```

## Compute all error metrics

```python
# Compute error metrics for each model
metrics_langmuir = all_error_metrics(q_exp, q_pred_langmuir, len(popt_langmuir))
metrics_sips = all_error_metrics(q_exp, q_pred_sips, len(popt_sips))
metrics_freundlich = all_error_metrics(q_exp, q_pred_freundlich, len(popt_freundlich))

# Create DataFrame for display
df_metrics = pd.DataFrame({
    'Langmuir': metrics_langmuir,
    'Sips': metrics_sips,
    'Freundlich': metrics_freundlich
})

print("Error Metrics Comparison:")
print(df_metrics.round(4))
```

## Information criteria and Akaike weights

```python
# Compute AICc and BIC for each model
n_points = len(q_exp)

aicc_langmuir = compute_AICc(q_exp, q_pred_langmuir, len(popt_langmuir))
aicc_sips = compute_AICc(q_exp, q_pred_sips, len(popt_sips))
aicc_freundlich = compute_AICc(q_exp, q_pred_freundlich, len(popt_freundlich))

bic_langmuir = compute_BIC(q_exp, q_pred_langmuir, len(popt_langmuir))
bic_sips = compute_BIC(q_exp, q_pred_sips, len(popt_sips))
bic_freundlich = compute_BIC(q_exp, q_pred_freundlich, len(popt_freundlich))

# Compute Akaike weights
aicc_values = [aicc_langmuir, aicc_sips, aicc_freundlich]
weights = akaike_weights(aicc_values)

# Compute delta AICc
min_aicc = min(aicc_values)
delta_aicc = [aic - min_aicc for aic in aicc_values]

# Create DataFrame for display
df_ic = pd.DataFrame({
    'Model': ['Langmuir', 'Sips', 'Freundlich'],
    'AICc': aicc_values,
    'BIC': [bic_langmuir, bic_sips, bic_freundlich],
    'delta_AICc': delta_aicc,
    'Akaike_weight': weights
})

print("Information Criteria and Model Selection:")
print(df_ic.round(4))
print("\nInterpretation:")
best_model_idx = np.argmin(aicc_values)
best_model = ['Langmuir', 'Sips', 'Freundlich'][best_model_idx]
print(f"Best model by AICc: {best_model} (weight={weights[best_model_idx]:.4f})")
print(f"Models with delta_AICc < 2 are considered to have substantial support.")
print(f"Models with delta_AICc > 10 have essentially no support.")
```

## F-test: Langmuir vs Sips

```python
# Compute SSE for both models
SSE_langmuir = np.sum((q_exp - q_pred_langmuir)**2)
SSE_sips = np.sum((q_exp - q_pred_sips)**2)

# Perform F-test (Langmuir is nested in Sips)
F_stat, F_crit, p_value = f_test(
    SSE_reduced=SSE_langmuir,
    p_reduced=2,
    SSE_full=SSE_sips,
    p_full=3,
    n=n_points
)

print("F-test: Langmuir (reduced) vs Sips (full)")
print(f"SSE_Langmuir = {SSE_langmuir:.4f}")
print(f"SSE_Sips = {SSE_sips:.4f}")
print(f"F-statistic = {F_stat:.4f}")
print(f"F-critical (alpha=0.05) = {F_crit:.4f}")
print(f"p-value = {p_value:.4e}")
print("\nConclusion:")
if p_value < 0.05:
    print("The additional parameter in Sips model is statistically significant (p < 0.05).")
    print("Reject the simpler Langmuir model in favor of Sips.")
else:
    print("The additional parameter in Sips model is NOT statistically significant (p >= 0.05).")
    print("The simpler Langmuir model is preferred.")
```
